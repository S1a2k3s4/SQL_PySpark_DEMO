### Column Filtering

```python
df_trimmed=df.select("title","imdb_rating","industry")
df_trimmed.show(3,truncate=False)
```

### Row Filtering

```python
df=df.filter((df["release_year"]>=2000) & (df["release_year"]<=2010))
display(df)
```

```python
from pyspark.sql.functions import col
```

```python
df=df.filter(col("release_year").between(2000,2010))
display(df)
```

```python
unique_industries=df.select("industry").distinct()
display(unique_industries)
```

```python
unique_lang=df.select("language").distinct()
display(unique_lang)
```

```python
df=df.withColumn('profit',col('revenue')-col('budget'))
display(df)
```

```python
df=df.withColumnRenamed('revenue','total_revenue')
df.printSchema()
```

```python
df=df.selectExpr("budget*12 as annual_budget").show()
```
